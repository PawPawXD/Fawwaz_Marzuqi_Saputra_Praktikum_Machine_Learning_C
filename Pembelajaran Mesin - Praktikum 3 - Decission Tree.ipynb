```python
#import library

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Pre Processing

```python
#fawwaz marzuqi saputra_202431035
df = pd.read_csv('iris-flower.csv', sep=",")
df
```

```python
df.info()
#fawwaz marzuqi saputra_202431035
```

```python
df['Id'].unique()
#fawwaz marzuqi saputra_202431035
```

```python
df['Species'].unique()
#fawwaz marzuqi saputra_202431035
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Species'] = le.fit_transform(df['Species'])
print(df['Species'].unique())

for  i, label in enumerate(le.classes_):
    print (f"{i} = {label}")
```

```python
df
```

## Modelling

```python
x = df.drop(['Id', 'Species'], axis=1)
x
```

```python
y = df['Species']
y
```

```python
#fawwaz marzuqi saputra_202431035
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=0)

from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier()
classifier.fit(x_train, y_train)
```

```python
#fawwaz marzuqi saputra_202431035
from sklearn import tree

plt.subplots(figsize = (15,19))
tree.plot_tree(classifier, fontsize = 10)
plt.show()
```

```python
#fawwaz marzuqi saputra_202431035
from sklearn.tree import DecisionTreeClassifier
classifier = DecisionTreeClassifier(criterion='entropy', max_depth=5, random_state=42)
classifier.fit(x_train, y_train)
```

```python
#fawwaz marzuqi saputra_202431035
from sklearn import tree

plt.subplots(figsize = (15,19))
tree.plot_tree(classifier, fontsize = 10)
plt.show()
```

```python
#fawwaz marzuqi saputra_202431035
depth = classifier.get_depth()
depth
```

```python
#evaluasi
#fawwaz marzuqi saputra_202431035

from sklearn.metrics import accuracy_score
y_pred = classifier.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)
print(akurasi)
```
